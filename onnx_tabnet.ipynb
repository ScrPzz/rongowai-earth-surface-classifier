```python
import torch
from pytorch_tabnet.tab_model import TabNetClassifier
import zipfile
import os

# Load TabNet model from the zip file
model_path = r"C:\Users\atogni\Desktop\rongowai\geo-k-compression_model\tabnet_binary_classifier_gpu.zip.zip"

model = TabNetClassifier()
model.load_model(model_path)
```

```python
dummy_input = torch.randn(3141, 88).float()
```

```python
torch.onnx.export(
    model, dummy_input, "tabnet.onnx",
    export_params=True,    # Store the trained parameters
    opset_version=11,      # Use ONNX opset 11+
    do_constant_folding=True,  # Optimize graph by folding constants
    input_names=["input"], output_names=["output"]
)
```
